```python
# 學生可以透過 prompt: "導入 numpy, matplotlib.pyplot，導入 matplotlib 與 fontManager 並設定中文字型支援，導入 scipy.cluster.hierarchy 的 linkage, dendrogram，以及 sklearn.cluster 的 AgglomerativeClustering，並生成範例數據集"
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib as mlp
from matplotlib.font_manager import fontManager

fontManager.addfont('ChineseFont.ttf')
mlp.rc('font', family='ChineseFont')

# 產生隨機範例資料
X, y = make_blobs(n_samples=50, centers=3, cluster_std=0.60, random_state=0)
```

```python
# 學生可以透過 prompt: "使用 scipy.cluster.hierarchy 的 linkage 函數計算連結矩陣（設定方法為 'ward'），並使用 dendrogram 繪製層次結構樹狀圖"
# 計算連結矩陣
Z = linkage(X, method='ward')

# 繪製樹狀圖
plt.figure(figsize=(10, 5))
dendrogram(Z)
plt.title('階層式集群樹狀圖 (Dendrogram)')
plt.xlabel('資料點索引')
plt.ylabel('距離 (Ward 距離)')
plt.show()
```

```python
# 學生可以透過 prompt: "使用 sklearn 的 AgglomerativeClustering 初始化階層集群模型（設定 n_clusters=3, linkage='ward'），擬合數據並預測類別標籤，以散佈圖進行分群結果視覺化"
# 初始化並訓練模型
agg_clustering = AgglomerativeClustering(n_clusters=3, linkage='ward')
y_pred = agg_clustering.fit_predict(X)

# 繪製分群結果
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_pred, s=50, cmap='viridis')
plt.title('階層式集群分群結果')
plt.xlabel('特徵 1')
plt.ylabel('特徵 2')
plt.show()
```

## 實作說明
- **資料產生**：使用 `make_blobs` 產生 50 個二維空間的範例資料點。
- **樹狀圖 (Dendrogram)**：使用 SciPy 的 `linkage` 與 `dendrogram` 函數繪製階層關係樹，幫助判斷最佳的分群數。
- **凝聚式階層集群**：使用 scikit-learn 的 `AgglomerativeClustering`，設定分群數為 3 進行擬合，並以散佈圖進行分群視覺化。
