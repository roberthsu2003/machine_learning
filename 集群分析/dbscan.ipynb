```python
# 設定中文字型支援
# --- 設定 Matplotlib 中文字型支援（Mac / Windows / Colab 通用）---
import glob
import os
import mplfonts
from matplotlib import font_manager
from mplfonts import use_font

for f in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(f)   # 直接註冊 mplfonts 內附的字型，不改任何檔案
use_font("Noto Sans CJK SC")              # 套用思源黑體，負號也會一起處理
```

```python
# =============================================================================
# 教學目標：產生雙半月形資料：K-means 做不來的形狀（動機格）
# 適用對象：機器學習初學者
# 學習重點：非線性月牙即 DBSCAN 的主場；noise 控制點的散開程度
# =============================================================================

# 學生可以透過 prompt: "導入 numpy, matplotlib.pyplot，導入 matplotlib 與 fontManager 並設定中文字型支援，導入 sklearn.datasets 的 make_moons，以及 sklearn.cluster 的 DBSCAN，並生成雙半月形範例數據集"
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import matplotlib as mlp


# 產生非線性分佈的雙半月形資料
X, y = make_moons(n_samples=200, noise=0.05, random_state=0)
```

```python
# =============================================================================
# 教學目標：用 DBSCAN(eps=0.2, min_samples=5) 擬合併數出噪聲點
# 適用對象：機器學習初學者
# 學習重點：核心點定義：半徑內夠多鄰居；標籤 -1 即密度不足的雜訊
# =============================================================================

# 學生可以透過 prompt: "使用 sklearn 的 DBSCAN 初始化密度分群模型（設定 eps=0.2, min_samples=5），訓練模型預測標籤，並印出偵測到的噪聲點數量"
# 初始化 DBSCAN 模型
dbscan = DBSCAN(eps=0.2, min_samples=5)
y_pred = dbscan.fit_predict(X)

# 計算噪聲點 (標籤為 -1 的點)
n_noise = list(y_pred).count(-1)
print(f"分群完成，偵測到的噪聲點數量: {n_noise}")
```

```python
# =============================================================================
# 教學目標：畫出漸層群集＋黑叉噪聲的分群圖（本檔結論）
# 適用對象：機器學習初學者
# 學習重點：與 K-means 圓形群對照：形狀不規則＋自動過濾正是密度法的價值
# =============================================================================

# 學生可以透過 prompt: "使用 matplotlib 繪製散佈圖展示 DBSCAN 分群結果，將非噪聲點以漸層色標記，將噪聲點單獨以黑色叉號 (x) 標記並印出圖表"
plt.figure(figsize=(8, 6))
# 繪製非噪聲點
plt.scatter(X[y_pred != -1, 0], X[y_pred != -1, 1], c=y_pred[y_pred != -1], s=50, cmap='viridis', label='群集')
# 繪製噪聲點
plt.scatter(X[y_pred == -1, 0], X[y_pred == -1, 1], c='black', marker='x', s=50, label='噪聲/異常值')
plt.title('DBSCAN 密度分群結果')
plt.xlabel('特徵 1')
plt.ylabel('特徵 2')
plt.legend()
plt.show()
```

## 實作說明
- **資料產生**：使用 `make_moons` 產生雙半月形的非線性數據，這種數據是傳統 K-means 無法良好分群的。
- **模型擬合**：DBSCAN 依據資料密度進行分群，`eps=0.2` 指定鄰域半徑，`min_samples=5` 為核心點所需的最少鄰近點數。
- **噪聲識別**：DBSCAN 能自動將密度不足的離群點標記為 `-1` (即噪聲/異常值)，在散佈圖中我們使用黑色 `x` 標識。
