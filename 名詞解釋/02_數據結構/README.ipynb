# 02. 數據結構 (Data Structures: Features & Labels)

本範例透過載入繁體中文「房產資料集」，帶領大家理解機器學習的基礎數據結構：
1. **特徵矩陣 $X$ (Feature Matrix)**：樣品的各項屬性（坪數、屋齡、捷運距離等）。
2. **標籤向量 $y$ (Target/Label Vector)**：目標變數（房屋總價）。
3. **視覺化圖表**：使用 Matplotlib 與 Seaborn 繪製繁體中文特徵散佈圖。

```python
# --- 設定 Matplotlib / Seaborn 中文字型支援（Mac / Windows / Colab 通用）---
import glob
import os
import mplfonts
from matplotlib import font_manager
import matplotlib.pyplot as plt
import seaborn as sns
from mplfonts import use_font

for f in glob.glob(os.path.join(os.path.dirname(mplfonts.__file__), "fonts", "*.[ot]tf")):
    font_manager.fontManager.addfont(f)
use_font("Noto Sans CJK SC")
plt.rcParams['axes.unicode_minus'] = False
print("繁體中文字型已成功載入！")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 讀取繁體中文來源資料集
df = pd.read_csv('房產資料集.csv')
print("=== 房產資料集預覽 ===")
display(df.head())
```

```python
# 2. 區分特徵矩陣 X 與標籤向量 y
X = df[['建築坪數', '屋齡_年', '行政區域', '捷運距離_公尺']]  # 特徵矩陣 (2D DataFrame)
y = df['房屋總價_萬元']                                   # 標籤向量 (1D Series)

print("特徵矩陣 X 的形狀 (樣本數, 特徵數):", X.shape)
print("標籤向量 y 的形狀 (樣本數,):", y.shape)
print("\n=== 特徵資料型態 ===")
print(X.dtypes)
```

```python
# 3. 繪製特徵與標籤的關係圖
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='建築坪數', y='房屋總價_萬元', hue='行政區域', s=120, palette='Set2')
plt.title('建築坪數與房屋總價關係圖', fontsize=14, pad=12)
plt.xlabel('建築坪數 (坪)', fontsize=12)
plt.ylabel('房屋總價 (萬元)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("💡 觀念總結：特徵矩陣 X 是模型觀察的世界，標籤 y 則是模型要預測的解答！")
```
