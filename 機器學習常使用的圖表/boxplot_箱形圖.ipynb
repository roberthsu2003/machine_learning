# 箱形圖 (Box Plot)

箱形圖（又稱盒鬚圖）是一種用於呈現數據分佈特徵的統計圖表。它能直觀地展示數據的：
- **中位數 (Median / Q2)**：數據排序後的中間值。
- **上下四分位數 (Q1, Q3)**：數據的第 25% 與第 75% 百分位數。
- **四分位距 (IQR)**：$Q3 - Q1$，即中間 50% 數據的跨度。
- **最大與最小值限制線 (Whiskers)**：通常設定在 $Q1 - 1.5 \times IQR$ 到 $Q3 + 1.5 \times IQR$ 的範圍。
- **離群值 (Outliers)**：超出限制線範圍的異常點（以獨立點表示）。

本篇將使用 `Seaborn` 繪製不同類別數據的箱形圖，並示範如何識別離群值。

## 2. 建立示範數據
我們模擬公司不同部門的薪資分佈，並手動加入幾個極端高薪（離群值）來展示箱形圖的診斷功能。

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# 模擬各部門的薪資（萬元/月）
rd_salary = np.random.normal(loc=8, scale=1.2, size=100)  # 研發部研發人員
marketing_salary = np.random.normal(loc=5.5, scale=0.8, size=100)  # 行銷部
hr_salary = np.random.normal(loc=4.8, scale=0.6, size=100)  # 人事部

# 故意在研發部中加入離群值 (極端高薪的高階主管)
rd_salary = np.append(rd_salary, [15.0, 16.5, 17.2])

# 建立 DataFrame
data = []
for s in rd_salary:
    data.append(["研發部", s])
for s in marketing_salary:
    data.append(["行銷部", s])
for s in hr_salary:
    data.append(["人事部", s])

df = pd.DataFrame(data, columns=["部門", "月薪(萬元)"])
df.sample(5)
```

## 3. 繪製箱形圖
我們使用 `Seaborn` 的 `boxplot` 繪製各部門薪資的箱形圖，觀察中位數、跨度與離群值。

```python
import seaborn as sns

plt.figure(figsize=(8, 6))

# 繪製箱形圖
sns.boxplot(data=df, x="部門", y="月薪(萬元)", palette="Set3")

# 可以搭配條狀分布點 (stripplot) 來觀察實際數據點的分布（可選）
sns.stripplot(data=df, x="部門", y="月薪(萬元)", color="black", size=3, alpha=0.3, jitter=0.2)

plt.title("各部門月薪分佈與離群值診斷", fontsize=14)
plt.xlabel("部門", fontsize=12)
plt.ylabel("月薪 (萬元)", fontsize=12)
plt.grid(True, axis='y', linestyle="--", alpha=0.5)
plt.show()
```
