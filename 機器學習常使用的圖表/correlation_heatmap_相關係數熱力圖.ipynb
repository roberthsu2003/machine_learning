# 特徵相關係數熱力圖 (Correlation Heatmap)

在特徵工程與探索性數據分析 (EDA) 中，理解變數兩兩之間的**線性相關程度**至關重要。這有助於我們：
- 識別哪些特徵對預測目標有強烈影響。
- 篩選掉高度共線性 (Multicollinearity) 的特徵，避免模型（如線性迴歸）產生過擬合或不穩定的權重。

本篇將使用 `Pandas` 計算皮爾森相關係數矩陣，並利用 `Seaborn` 繪製熱力圖。

## 1. 環境設定與下載中文字型
在 Google Colab 中繪圖若使用中文標籤常會出現亂碼（豆腐塊），我們需要下載並載入中文字型。

```python
# 下載中文字型
import os
import wget

if not os.path.exists("ChineseFont.ttf"):
    wget.download("https://github.com/roberthsu2003/machine_learning/raw/refs/heads/main/source_data/ChineseFont.ttf")
```

```python
# 設定 Matplotlib 使用下載的中文字型
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.font_manager import fontManager

fontManager.addfont("ChineseFont.ttf")
mpl.rc('font', family="ChineseFont")
mpl.rcParams['axes.unicode_minus'] = False  # 正常顯示負號
```

## 2. 建立示範數據
我們建立一個包含生理特徵的數據集，特徵之間具有預設的物理關聯性。

```python
import numpy as np
import pandas as pd

np.random.seed(42)
n_samples = 200

# 模擬身高與體重的關聯
height = np.random.normal(170, 8, n_samples)
weight = height * 0.4 + np.random.normal(0, 5, n_samples)

# 模擬年齡
age = np.random.randint(18, 65, n_samples)

# 模擬每日運動時間 (小時) 與體脂肪率的關聯 (運動時間越長，體脂通常較低)
exercise_time = np.random.uniform(0.5, 3.0, n_samples)
body_fat = 35 - exercise_time * 4 - (height - 170) * 0.1 + np.random.normal(0, 3, n_samples)

# 建立 DataFrame
df = pd.DataFrame({
    "身高(cm)": height,
    "體重(kg)": weight,
    "年齡(歲)": age,
    "每日運動時間(hr)": exercise_time,
    "體脂肪率(%)": body_fat
})
df.head()
```

## 3. 計算相關係數矩陣
使用 `df.corr()` 可以快速計算特徵之間的皮爾森相關係數矩陣（數值介於 -1 到 1 之間）。

```python
corr_matrix = df.corr()
corr_matrix
```

## 4. 繪製熱力圖 (Heatmap)
我們使用 `Seaborn` 的 `heatmap` 函數進行視覺化，讓相關性的強弱一目了然。

```python
import seaborn as sns

plt.figure(figsize=(8, 6))

# 繪製熱力圖
# annot=True 顯示數值，cmap 使用發散型色盤，vmin/vmax 設定色帶範圍
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f", linewidths=0.5)

plt.title("生理指標特徵相關係數熱力圖", fontsize=14)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
