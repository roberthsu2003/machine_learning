```python
import requests

url = 'https://notify-api.line.me/api/notify'
token = '2pWhgX4pM4hImT4faMWi3ZDXCJMM06RKTSgIqQ6e649'
headers = {
    'Authorization': 'Bearer ' + token    # 設定權杖
}
data = {
    'message':'測試一下！'     # 設定要發送的訊息
}
data = requests.post(url, headers=headers, data=data)   # 使用 POST 方法
```

##實作FinMindApi爬取個股資訊

```python
from FinMind.data import DataLoader
import requests
import numpy as np
import datetime

# 獲取當前日期
today = datetime.date.today()

# 從使用者處獲取股票代碼
stock_id = input("請輸入欲查詢的股票代碼: ")

# 創建 DataLoader 實例
api = DataLoader()

# 使用 DataLoader 獲取指定日期的台灣股票日數據
df = api.taiwan_stock_daily(
    stock_id=stock_id,
    start_date=today.strftime('%Y-%m-%d'),
    end_date=today.strftime('%Y-%m-%d')
)

# 將 DataFrame 轉換為 numpy 陣列
tmp_msg = np.asarray(df)

# 構建發送到 LINE Notify 的訊息內容
msg = '\n' + \
      '今天日期：' + str(tmp_msg[0][0]) + '\n' + \
      '股票代號：' + str(tmp_msg[0][1]) + '\n' + \
      '成交量：' + str(tmp_msg[0][2]) + '\n' + \
      '成交金額：' + str(tmp_msg[0][7]) + '\n'

#print(msg)

def lineNotifyMessage(token, msg):
    """發送 LINE Notify 訊息"""

    # 設定 HTTP 標頭
    headers = {
        "Authorization": "Bearer " + token,
        "Content-Type" : "application/x-www-form-urlencoded"
    }

    # 設定負載
    payload = {'message': msg }

    # 發送 POST 請求
    r = requests.post("https://notify-api.line.me/api/notify", headers = headers, params = payload)

    return r.status_code

if __name__ == "__main__":
    token = '2pWhgX4pM4hImT4faMWi3ZDXCJMM06RKTSgIqQ6e649'  # 在這裡輸入你的 LINE Notify 權杖
    message = msg
    lineNotifyMessage(token, message)
```
