```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 設定爬蟲目標網址
url = "https://www.bot.com.tw/web/services/ef_gold/efg_dailyquote/yg_dailyquote.asp"

# 設定爬蟲參數
payload = {
    'ST_DATE': '2022/03/22',  # 起始日期
    'ED_DATE': '2023/03/22',  # 結束日期
    'query': 'query'  # 固定參數
}

# 使用 POST 方法發送請求
r = requests.post(url, data=payload)

# 解析 HTML 資料
soup = BeautifulSoup(r.text, 'html.parser')

# 找出表格
table = soup.find('table', class_='table_normal')
rows = table.find_all('tr')

# 提取表格中的資料
data = []
for row in rows[1:]:
    cols = row.find_all('td')
    cols = [col.text.strip() for col in cols]
    data.append(cols)

# 將資料轉成 DataFrame
df = pd.DataFrame(data, columns=['日期', '買進金額', '賣出金額'])

# 將日期欄位轉成日期格式
df['日期'] = pd.to_datetime(df['日期'], format='%Y/%m/%d')

# 將金額欄位轉成數值格式
df['買進金額'] = pd.to_numeric(df['買進金額'].str.replace(',', ''))
df['賣出金額'] = pd.to_numeric(df['賣出金額'].str.replace(',', ''))

# 輸出結果
print(df)
```
