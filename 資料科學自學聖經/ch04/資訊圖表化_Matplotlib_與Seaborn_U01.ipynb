## [語法] plt.plot(listx, listy)

※[Note]: X軸串列與Y軸串列的元素數目必須相同!!

4.2.1 繪製折線圖

```python
import matplotlib.pyplot as plt

listx = [1,5,7,9,13,16]
listy = [15,50,80,40,70,50]
plt.plot(listx, listy)
plt.show()
```

4.2.2 設定線條、標記及圖例

[Note]
1.linewidth(lw):設定線條寬度,預設為: 1.0 \
2.color:設定線條顏色,預設為:blue \
3.linestyle(ls):預設值有:實線(-), 其他形式有:虛線(--),虛點線(-.), 點線(:) \
4.marker:設定標記樣式 \
5.markersize(ms):標記大小 \
6.組合字串: color, linestyle, marker可直接合併設定, \
例如: 設定 綠色, 虛線, "*"狀標記 \
 plt.plot(listx, listy, 'g--*')  \
7.label:設定圖例名稱, 此屬性需搭配 legend()才有效果\

```python
import matplotlib.pyplot as plt

listx = [1,5,7,9,13,16]
listy = [15,50,80,40,70,50]
plt.plot(listx, listy, color="red", lw="2.0", ls="-.", label="label")
plt.legend()
plt.show()
```

4.2.3 設定圖表及XY軸標題

```python
import matplotlib.pyplot as plt

listx = [1,5,7,9,13,16]
listy = [15,50,80,40,70,50]
plt.plot(listx, listy, color="blue", lw="2.0", ls="--", label="label")
plt.legend()
plt.title("Chart Title", fontsize=20)  # 圖表標題
plt.xlabel("X-Label", fontsize=14)    # x軸標題
plt.ylabel("Y-Label", fontsize=14)    # y軸標題
plt.show()
```

## 4.3 長條圖(bar)與橫條圖(barh)

4.3.1 繪製長條圖 \
[語法] plt.bar(x軸串列, y軸串列, width=0.8, bottom=0[, 其他參數]) \

width:每個矩型的寬度,以百分比為單位設定,預設值為:0.8。 \
bottom:設定每個矩形y軸的起始位置,預設值為:0。 \
color:在Python的預設顏色為: \
  b: blue, g: green,  r: red \
  c: cyan, m: magenta, y: yellow \
  k: black, w: white \
label:設定每個圖例名稱,需要搭配 legend()

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager

# 加入中文字型設定：翰字鑄造-台北黑體
fontManager.addfont('/content/TaipeiSansTCBeta-Regular.ttf')
matplotlib.rc('font', family='Taipei Sans TC Beta')

#fontManager.addfont('/content/TaipeiSansTCBeta-Regular.ttf')
#matplotlib.rc('font', family='Noto Sans TC')

listx = ['c','c++','c#','java','python']
listy = [45,28,38,32,50]
plt.bar(listx, listy, width=0.5, color=['r','g','c','m','y'])
plt.title("資訊程式課程選修人數")
plt.xlabel("程式課程")
plt.ylabel("選修人數")
plt.show()
```

4.3.2 繪製橫條圖 \
[語法] plt.barh(y軸串列, x軸串列, heigth=0.8, left=0[, 其他參數]) \
y軸串列:顯示每個項目的名稱串列或是序列串列 \
x軸串列:顯示每個項目的數值串列 \


```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager
# 加入中文字型設定：翰字鑄造-台北黑體
fontManager.addfont('/content/TaipeiSansTCBeta-Regular.ttf')
matplotlib.rc('font', family='Taipei Sans TC Beta')

#fontManager.addfont('/content/static/NotoSansTC-Regular.ttf')
#matplotlib.rc('font', family='Noto Sans TC')

listy = ['c','c++','c#','java','python']
listx = [45,28,38,32,50]
plt.barh(listy, listx, height=0.5, color=['r','g','b'])
plt.title("資訊程式課程選修人數")
plt.xlabel("程式課程")
plt.ylabel("選修人數")
plt.show()
```

4.3.3 堆疊長條圖 \
繪製堆疊長條圖時,需用到bottom屬性。\
Ex: plt.bar(listx, listy2, width=0.5, bottom=listy1, label='女') \

其中 bottom=listy1 表示在繪製第二組長條圖時,將y軸的起點設定為第一組資料的y軸高度。

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager
fontManager.addfont('/content/static/NotoSansTC-Regular.ttf')
matplotlib.rc('font', family='Noto Sans TC')

listx = ['c','c++','c#','java','python']
listy1 = [25,20,20,16,28]
listy2 = [20,8,18,16,22]
plt.bar(listx, listy1, width=0.5, label='男')
plt.bar(listx, listy2, width=0.5, bottom=listy1, label='女')
plt.legend()
plt.title("資訊程式課程選修人數")
plt.xlabel("程式課程")
plt.ylabel("選修人數")
plt.show()
```

4.3.4 繪製並列長條圖 \
由於並列長條圖每個X軸的刻度間距是相等的,當繪製每個項目的矩形時,預設會以刻度為寬度的中心點,因此每個項目中子項目在刻度上的起始位置就非常重要。

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager
fontManager.addfont('/content/static/NotoSansTC-Regular.ttf')
matplotlib.rc('font', family='Noto Sans TC')

width = 0.25
listx = ['c','c++','c#','java','python']
listx1 = [x - width/2 for x in range(len(listx))] #使用串列綜合表達式的函式,將子項目串列中的值,逐一向左運算
listx2 = [x + width/2 for x in range(len(listx))] #使用串列綜合表達式的函式,將子項目串列中的值,逐一向右運算
listy1 = [25,20,20,16,28]
listy2 = [20,8,18,16,22]
plt.bar(listx1, listy1, width, label='男')
plt.bar(listx2, listy2, width, label='女')
plt.xticks(range(len(listx)), labels=listx)
plt.legend()
plt.show()
```

## 4.4 圓形圖(Pie)

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.font_manager import fontManager
fontManager.addfont('/content/static/NotoSansTC-Regular.ttf')
matplotlib.rc('font', family='Noto Sans TC')

sizes = [25, 30, 15, 10]
labels = ["北部", "西部", "南部", "東部"]
colors = ["red", "green", "blue", "yellow"]
explode = (0.2, 0, 0, 0)  #凸出的項目 0:表示正常顯示
plt.pie(sizes,
	explode = explode,
	labels = labels,
	colors = colors,
	labeldistance = 1.1,   #項目標題與圓心的距離是半徑的多少倍
	autopct = "%2.1f%%",   #項目百分比格式,語法為:[%格式%%]
	pctdistance = 0.6,    #百分比文字與圓心的距離是半徑的多少倍
	shadow = True,      #True:有陰影  False:沒有
	startangle = 90)     #開始繪圖的起始角度
plt.show()
```

## 4.5 直方圖(hist)

## 4.6 散佈圖(scatter)
[語法] plt.scatter(x軸串列, y軸串列[,其他參數]) \

x軸串列, y軸串列:是長度相同的陣列或串列,為必要參數。\
s:標記大小,可以是數值或數值串列。\
c:標記顏色。\
marker:標記樣式,預設值:'o'。\
alpha:標記的透明度,值在0-1,預設值:None(不透明)。


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [1, 4, 9, 16, 7, 15, 17, 19]
sizes = [20, 200, 100, 50, 500, 1000, 60, 90]
colors = ["red","green","black","orange","purple","pink","cyan","magenta"]
plt.scatter(x, y, s=sizes, c=colors)
plt.show()
```

## Lab4-2.承上圖,如何在圖中央繪製十字線?

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [1, 4, 9, 16, 7, 15, 17, 19]
sizes = [20, 200, 100, 50, 500, 1000, 60, 90]
colors = ["red","green","black","orange","purple","pink","cyan","magenta"]
#繪製圖表區

#繪製十字線
plt.axvline(x=4.5, color='red', linestyle='--', linewidth=1) #繪製橫線
plt.axhline(y=10, color='red', linestyle='--', linewidth=1) #繪製直線

plt.scatter(x, y, s=sizes, c=colors)

plt.show()
```

## 4.8 設定圖表區(figure)

[語法] plt.figure([設定屬性參數]) \
常用參數: \
figsize:設定方式為串列:[寬,高],單位為英吋,預設值:[6.4, 4.8]\
dpi:設定解析度(dpi:dot per inch)。\
facecolor:設定背景色,預設值:white。\
edgecolor:設定邊線顏色,預設值:white。\
frameon:布林值,設定是否有邊框,預設值:True。

```python
import matplotlib.pyplot as plt
# 新增圖表區
plt.figure()
plt.plot([1,2,3])
# 新增圖表區並設定屬性
plt.figure(
    figsize=[10,4],
    facecolor="whitesmoke",
    edgecolor="b",
    linewidth=10,
    frameon=True)
plt.plot([1,2,3])
plt.show()
```

## 4.9 在圖表區加入多張圖表(subplot, axes)

4.9.1 用欄列排列多張圖表: subplot() \
[語法] plt.subplot(橫列數, 直欄數, 圖表索引值) \

[Ex1]在圖表區加入2列1欄的二張圖表:

```python
import matplotlib.pyplot as plt
plt.figure(figsize=[8,8])
plt.subplot(2,1,1)
plt.title(label='Chart 1', fontsize=20)
plt.plot([1,2,3],'r:o')

plt.subplot(2,1,2)
plt.title(label='Chart 2', fontsize=20)
plt.plot([1,2,3],'g--o')
plt.show()
```

[Ex2]在圖表區加入1列2欄的二張圖表:

```python
import matplotlib.pyplot as plt
plt.figure(figsize=[8,8])
plt.subplot(1,2,1)
plt.title(label='Chart 1', fontsize=20)
plt.plot([1,2,3],'r:o')

plt.subplot(1,2,2)
plt.title(label='Chart 2', fontsize=20)
plt.plot([1,2,3],'g--o')
plt.show()
```

[Ex3]在圖表區加入2列2欄的四張圖表:

```python
import matplotlib.pyplot as plt
plt.figure(figsize=[8,8])
plt.subplot(2,2,1)
plt.title(label='Chart 1')
plt.plot([1,2,3],'r:o')
plt.subplot(2,2,2)
plt.title(label='Chart 2')
plt.plot([1,2,3],'g--o')
plt.subplot(2,2,3)
plt.title(label='Chart 3')
plt.plot([1,2,3],'b:o')
plt.subplot(2,2,4)
plt.title(label='Chart 4')
plt.plot([1,2,3],'y--o')
plt.show()
```

4.9.2 用相對位置排列多張圖表: axes \
[語法] plt.axes([與左邊界距離, 與下邊界距離, 寬, 高]) \
plt.axes()是以圖表區的左下角為原點。在語法中的四個參數值,皆以0-1的浮點數來計算。

## Lab4-3.繪製折線圖的趨勢線

[提示] \
1.從random模組載入 randrange()以快速生成一些資料。\
2.使用NumPy套件的 polyfit() 和 poly1d()，繪製趨勢線。

製作人:00_老趙
檔名:Lab1030-1

```python
import numpy as np
import random as rd
from matplotlib import pyplot as plt

# 加入中文字型設定：翰字鑄造-台北黑體
fontManager.addfont('/content/TaipeiSansTCBeta-Regular.ttf')
matplotlib.rc('font', family='Taipei Sans TC Beta')
#fontManager.addfont('/content/static/NotoSansTC-Regular.ttf')

x = np.array([i for i in range(1, 11)])
y = np.array([rd.randrange(10, 100) for _ in range(10)])  # 生成隨機的y軸數據

# Plot the Data itself.
plt.plot(x, y)

# 使用polyfit()擬合趨勢線
degree = 1  # 多项式的次数，这里是一次线性拟合
coefficients = np.polyfit(x, y, degree)
poly = np.poly1d(coefficients)

# 计算R平方值
y_fit = poly(x)  # 使用拟合线计算y值
residuals = y - y_fit
ss_res = np.sum(residuals**2)
ss_tot = np.sum((y - np.mean(y))**2)
r_squared = 1 - (ss_res / ss_tot)

# 繪製趨勢線
x_fit = np.linspace(min(x), max(x), 100)  # 生成用于擬合線的x軸數據
y_fit = poly(x_fit)  # 计算擬合線的y軸數據
plt.plot(x_fit, y_fit, label=f'趨勢線: y = {coefficients[0]:.2f}x + {coefficients[1]:.2f}', color='r')

# 顯示R平方值
plt.text(2, 90, f'R平方值 = {r_squared:.4f}', fontsize=10, color='k')

def reg(x,y):
    coefficients = np.polyfit(x,y,1) # 利用 polyfit 幫我們算出資料 一階擬合的 a, b 參數
    p = np.poly1d(coefficients) # 做出公式, print 的結果是 coefficients[0] * X + coefficients[1]
    #coefficient_of_dermination = r2_score(y, p(x)) // 計算相關係數用，這裡沒有用到
    return coefficients, p

(arg1, arg2), text1 = reg(x,y)
#arg1, arg2 用來承接 coefficients[0], coefficients[1], text1 承接 p

trend_line = x*arg1 + arg2

# Display the Trendline
plt.plot()
plt.plot(trend_line) #放上趨勢線圖
plt.text(50, 10, text1, fontsize=14) #放上趨勢線公式

plt.show()
```

2.簡單熱力圖  
使用np函數設定隨機種子，同樣也是使用np函數，從均勻分配[0~1）之間抽出2*3陣列。要特別注意，這裡抽取的範圍包含0，但不包含1。接下來使用heatmap函數繪製出熱力圖，同時在最後一行埶行uniform_data，可以顯示出隨機抽取2*3的數值。

Ex.2-1下面的熱力圖，就x軸而言，數值是從0到2，y軸是1到0。最右邊的長條圖，最高數值0.7以上是乳白色，最下面黑色是0.45，中間0.55的部分是用紫色來表示。    
數據第一個column是0.5488135及0.54488318，接近0.55，因此熱力圖最左邊二個都是紫色。

數據第二個column是0.71518937及0.4236548，因此熱力圖中間第一個接近乳白色，第二個接近黑色。

數據第三個column是0.60276338及0.64589411，因此熱力圖最右邊第一個接近橘色，第二個顏色則是在橘色與乳白色之間。

```python
#設定隨機種子
np.random.seed(0)

#從均勻分配[0~1)之間抽出2*3陣列
uniform_data=np.random.rand(2,3)

sns.heatmap(uniform_data)

uniform_data
```

Ex2-2.若使用10*10均勻分配資料，這次就x軸而言，數值是從0到9，y軸是9到0。最右邊的長條圖，最高數值0.8以上是乳白色，最下面黑色數值約0.1，中間0.5的部分是用紫色來表示。可以看到數值愈高顏色愈淺，而數值愈低顏色愈深。

```python
#使用10*10均勻分配資料
np.random.seed(0)

uniform_data=np.random.rand(10,10)

sns.heatmap(uniform_data)
```

3. 顏色範圍  
一樣使用前面10*10均勻分配資料，數值範圍從0.5~0.8，這代表右邊長條的範圍，同樣是數值愈大則顏色愈淺。

```python
sns.heatmap(uniform_data, center=0.5)
```

在EX4-1.這個範例裡，使用np函數抽取標準常態分配10*10的數據。所謂標準常態分配是指平均數為0，標準差為1的常態分配。在此圖中，以零為中心點，因此中心點0為黑色，0以下的地方開始逐漸轉為淺藍色，而0以上開始逐漸轉為淡橘色。

常態分配圖 (引用:https://ntpuccw.blog/python-in-learning/lesson-5-probability-distributions-and-the-shapes/)

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

# Broadcasting 示範
fig, ax = plt.subplots()
xlim = [-15, 15]
mu = 0
s = np.arange(1, 6)
x = np.linspace(xlim[0], xlim[1], 1000)
Y = norm.pdf(x.reshape(-1, 1), mu, s)
# print(y.shape)
label = ["$\sigma$={}".format(i) for i in s] #  list comprehension (loop in bracket)
plt.plot(x, Y, label=label)
plt.legend()
plt.show()
```
